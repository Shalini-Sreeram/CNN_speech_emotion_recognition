```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import seaborn as sns
import cv2  
from moviepy.editor import VideoFileClip
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
```

```python
# Directory containing video files
video_directory = r'C:\Users\T 480\MALD\MELD\train\train_splits'

# Verify that the directory exists
if not os.path.exists(video_directory):
    raise ValueError(f"Directory not found: {video_directory}")

# List all mp4 video files
video_files = [f for f in os.listdir(video_directory) if f.endswith('.mp4')]
print(f"Found {len(video_files)} video files.")
```

```python
def get_video_metadata(video_path):
    try:
        with VideoFileClip(video_path) as video:
            duration = video.duration
            fps = video.fps
            resolution = video.size
            return duration, resolution[0], resolution[1], fps
    except Exception as e:
        print(f"Error retrieving metadata for {video_path}: {str(e)}")
        return None
```

```python
def extract_brightness(video_path):
    try:
        with VideoFileClip(video_path) as video:
            # Removed the `progress_bar` argument
            frames = [frame.mean() for frame in video.iter_frames(dtype='uint8')]
            average_brightness = sum(frames) / len(frames) if frames else 0
            return average_brightness
    except Exception as e:
        print(f"Error calculating brightness for {video_path}: {str(e)}")
        return None
```

```python
# Process the videos and store the results
results = []
for idx, file in enumerate(video_files):
    full_path = os.path.join(video_directory, file)
    metadata = get_video_metadata(full_path)
    if metadata:
        duration, width, height, fps = metadata
        brightness = extract_brightness(full_path)
        if brightness is not None:
            results.append([file, duration, width, height, fps, brightness])
    else:
        print(f"Skipping file {file} due to retrieval errors.")

    if (idx + 1) % 100 == 0 or idx == len(video_files) - 1:
        print(f"Processed {idx + 1}/{len(video_files)} files.")

# Create a DataFrame from the collected data
df_videos = pd.DataFrame(results, columns=['Filename', 'Duration', 'Width', 'Height', 'FPS', 'Average_Brightness'])
print(df_videos.head())
```

```python

import os
from moviepy.editor import VideoFileClip
import librosa
import numpy as np
# Function to extract audio from video
def extract_audio_from_video(video_path, output_audio_path):
    try:
        video = VideoFileClip(video_path)
        audio = video.audio
        audio.write_audiofile(output_audio_path, codec='pcm_s16le')  # WAV format
        print(f"Audio extracted and saved to {output_audio_path}")
        return True
    except Exception as e:
        print(f"Error extracting audio from {video_path}: {str(e)}")
        return False

# Function to extract audio features using Librosa
def extract_audio_features(audio_path):
    try:
        y, sr = librosa.load(audio_path, sr=None)
        mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13)
        mfccs_processed = np.mean(mfccs.T, axis=0)  # Taking the mean across time
        return mfccs_processed
    except Exception as e:
        print(f"Error extracting features from {audio_path}: {str(e)}")
        return None
```

```python
import os
import csv
import numpy as np

# Directory paths
video_directory = r'C:\Users\T 480\MALD\MELD\train\train_splits'
output_directory = r'C:\Users\T 480\MALD\MELD\train\audio_files'
os.makedirs(output_directory, exist_ok=True)

# Process all video files in the directory
video_files = [f for f in os.listdir(video_directory) if f.endswith('.mp4')]
audio_features_list = []

for file in video_files:
    video_path = os.path.join(video_directory, file)
    audio_output_path = os.path.join(output_directory, file.replace('.mp4', '.wav'))
    
    # Extract audio
    if extract_audio_from_video(video_path, audio_output_path):
        # Extract audio features
        audio_features = extract_audio_features(audio_output_path)
        if audio_features is not None:
            audio_features_list.append((file, audio_features))
            print(f"Extracted Audio Features for {audio_output_path}: {audio_features}")

# Save audio features to CSV file
csv_output_path = os.path.join(output_directory, 'audio_features.csv')
with open(csv_output_path, mode='w', newline='') as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(['File Name', 'Audio Features'])
    for file_name, features in audio_features_list:
        writer.writerow([file_name, features.tolist()])

print("Audio features saved to CSV.")
```
